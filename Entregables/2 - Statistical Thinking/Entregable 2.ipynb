7 - Delimitación del alcance del análisis (filtro de barrios a analizar) y eliminación de Outliers

```python
df_idporbarrio = df_1.groupby("neighbourhood_group", as_index=False)["id"].count()
df_idporbarrio["porcentaje"] = df_idporbarrio["id"] / df_idporbarrio["id"].sum() * 100
df_idporbarrio
```

```python
df_barrios_alcance = df_idporbarrio[df_idporbarrio["porcentaje"] > 3] 
df_barrios_alcance
```

```python
df_1_ampliado = (
    df_1.assign(
        barrio_alcance=lambda x: x["neighbourhood_group"].isin(df_barrios_alcance["neighbourhood_group"])
    ))
df_1_ampliado
```

```python
df_1_ampliado = df_1_ampliado.merge(
    df_idporbarrio[["neighbourhood_group", "porcentaje"]],
    on="neighbourhood_group",
    how="left"
)
```

```python
df_sin_outliers = df_1_ampliado.groupby("neighbourhood_group").apply(
    lambda x: x[
        (x["price"] >= x["price"].quantile(0.25) - 1.5 * (x["price"].quantile(0.75) - x["price"].quantile(0.25))) &
        (x["price"] <= x["price"].quantile(0.75) + 1.5 * (x["price"].quantile(0.75) - x["price"].quantile(0.25)))
    ]
).reset_index(drop=True)
df_sin_outliers
```

```python
sns.boxplot(
    data=df_sin_outliers,
    y="neighbourhood_group",
    x="price",
);
```

```python
sns.barplot(
    data=df_sin_outliers,
    y="neighbourhood_group",
    x="porcentaje",
    hue="barrios_alcance",
);
```

8 - Hipotesis 1 a analizar (Amenities): 

```python
amenities = ["amenities_wifi_internet", "amenities_air_conditioning", "amenities_heating", "amenities_tv"]
```

```python
df_filtrado = df_sin_outliers[df_sin_outliers["barrios_alcance"]]
for amenitie in amenities:
    plt.figure(figsize=(12,6))
    sns.boxplot(
        data=df_filtrado,
        x="neighbourhood_group",
        y="price",
        hue=amenitie
    )
    plt.title(f"Impacto de {amenitie} sobre el precio por barrio")
    plt.legend(title=amenitie)
    plt.show()
```

9 - Hipotesis 2 a analizar (Conditions): 

```python
conditions_discretas = ["cancellation_policy", "require_guest_profile_picture"]

```

```python
df_ = df_filtrado[df_filtrado['price'] < df_filtrado['price'].quantile(0.99)]

barrios = df_['neighbourhood_group'].unique()

for col in conditions_discretas:
    for barrio in barrios:
        subset = df_[df_['neighbourhood_group'] == barrio]
        plt.figure(figsize=(8,5))
        sns.boxplot(data=subset, x=col, y="price", palette="pastel")
        plt.title(f"Precio según {col} en {barrio}")
        plt.ylim(0, subset['price'].quantile(0.95))
        plt.ylabel("Precio por noche")
        plt.xlabel(col)
        plt.show()
```
