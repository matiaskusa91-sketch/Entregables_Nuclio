# Enunciado

Trabajas en una agencia que se dedica a organizar viajes a personas, tu cargo específicamente es analista de datos del área de clientes VIP.
Ha llegado a la agencia un cantante famoso, conocido en España, que desea que le ubiquéis un buen alojamiento que le permita disfrutar del sol para sus próximas vacaciones de verano que serán a finales del mes de julio del 2025.
Este cliente quiere descansar porque necesita prepararse para una gira de conciertos que dará en septiembre y por ahora quiere alejarse del foco mediático todo lo que pueda durante sus vacaciones, no le gustaría llamar la atención durante esos días ya que estará con sus 2 padres, sus 2 hermanos pequeños, su novia y los padres de su novia.
El cliente no tiene claro el destino al que quiere ir, solo sabe que tiene que ser en alguna playa. Tu trabajo será encontrar la propiedad en la que el cliente y su familia se alojarán durante sus vacaciones de verano teniendo en cuenta las exigencias del cliente.

# Primera Parte

### Primer paso: crear la query de cada uno de las exigencias del cliente. Se debe crear una query para cada punto. (7 queries en total)

-El cliente exige terraza y aire acondicionado, pero no quiere piscina.

```python
sSQL = """
SELECT * 
FROM alojamiento
WHERE TERRAZA = "SI" 
AND AIREC_ACOND = "SI" 
AND PISCINA = "NO";
 """
pd.read_sql(text(sSQL), conn)
```

-No se alojará en estudios que no tengan salón ni en propiedades con solo un baño

```python
sSQL = """
SELECT *
FROM alojamiento
WHERE SALON = "SI"
AND CANT_BANYOS > 1;
 """
pd.read_sql(text(sSQL), conn)
```

-La propiedad debe tener como mínimo 80m2

```python
sSQL = """
SELECT *
FROM alojamiento
WHERE SUPERFICIE >= 80;
 """
pd.read_sql(text(sSQL), conn)
```

-Como el cliente odia el ruido, no quiere alojarse en propiedades a menos de 1 km de una estación de metro ni a menos de 2 km del centro.


```python
sSQL = """
SELECT *
FROM ubicacion
WHERE dist_metro >= 1
AND dist_centro >=2;
 """
pd.read_sql(text(sSQL), conn)
```

```python
sSQL = """
SELECT *
FROM precio
WHERE precio_alquiler BETWEEN 1500 AND 2000;;
 """
pd.read_sql(text(sSQL), conn)
```

-Como en anteriores ocasiones ha tenido problemas de intentos de estafas con otras agencias, el cliente pide que el % de reserva no sea muy alto, pero tampoco tan bajo (porque le genera inseguridad) por lo que propone un 25%

### Segundo paso: crear en una única query la combinación de los filtros anteriores para hallar las propiedades que finalmente cumplen con todas las exigencias del cliente.

```python
sSQL = """
SELECT *
FROM (
    SELECT 
        A.CODIGO_ALOJAMIENTO,
        A.NOMBRE,
        A.CANT_HAB,
        A.CANT_BANYOS,
        A.SUPERFICIE,
        A.TERRAZA,
        A.AIREC_ACOND,
        A.PISCINA,
        A.SALON,
        U.PAIS,
        U.CIUDAD,
        U.DIST_METRO,
        U.DIST_CENTRO,
        PR.PRECIO_ALQUILER,
        PR.PORCENTAJE_RESERVA,
        PU.PUNTOS,
        PU.PUNTOS_AGENCIA
    FROM ALOJAMIENTO AS A
    INNER JOIN PRECIO      AS PR ON A.CODIGO_ALOJAMIENTO = PR.CODIGO_ALOJAMIENTO
    INNER JOIN PUNTUACION  AS PU ON A.CODIGO_ALOJAMIENTO = PU.CODIGO_ALOJAMIENTO
    INNER JOIN UBICACION   AS U  ON A.CODIGO_ALOJAMIENTO = U.CODIGO_ALOJAMIENTO
) AS ALOJ
WHERE ALOJ.TERRAZA = 'SI'
  AND ALOJ.AIREC_ACOND = 'SI'
  AND ALOJ.PISCINA = 'NO'
  AND ALOJ.SALON = 'SI'
  AND ALOJ.CANT_BANYOS > 1
  AND ALOJ.SUPERFICIE >= 80
  AND ALOJ.DIST_METRO >= 1
  AND ALOJ.DIST_CENTRO >= 2
  AND ALOJ.PRECIO_ALQUILER BETWEEN 1500 AND 2000
  AND ALOJ.PORCENTAJE_RESERVA = 25
  AND ALOJ.PUNTOS_AGENCIA > 4.5
  AND ALOJ.PUNTOS > 4;

 ;
 """
pd.read_sql(text(sSQL), conn)
```

### Tercer paso: Como habéis visto, tenemos 4 propiedades que cumplen con todas las exigencias de nuestro cliente, ahora bien, desde tu punto de vista, haciendo un análisis de los datos del cliente y de todas sus exigencias ¿Cuál de estos 4 alojamientos le recomendarías y por qué?

```python
sSQL = """
SELECT *
FROM (SELECT *
FROM (
    SELECT 
        A.CODIGO_ALOJAMIENTO,
        A.NOMBRE,
        A.CANT_HAB,
        A.CANT_BANYOS,
        A.SUPERFICIE,
        A.TERRAZA,
        A.AIREC_ACOND,
        A.PISCINA,
        A.SALON,
        U.PAIS,
        U.CIUDAD,
        U.DIST_METRO,
        U.DIST_CENTRO,
        PR.PRECIO_ALQUILER,
        PR.PORCENTAJE_RESERVA,
        PU.PUNTOS,
        PU.AGENCIA,
        PU.PUNTOS_AGENCIA
    FROM ALOJAMIENTO AS A
    INNER JOIN PRECIO      AS PR ON A.CODIGO_ALOJAMIENTO = PR.CODIGO_ALOJAMIENTO
    INNER JOIN PUNTUACION  AS PU ON A.CODIGO_ALOJAMIENTO = PU.CODIGO_ALOJAMIENTO
    INNER JOIN UBICACION   AS U  ON A.CODIGO_ALOJAMIENTO = U.CODIGO_ALOJAMIENTO
) AS ALOJ
WHERE ALOJ.TERRAZA = 'SI'
  AND ALOJ.AIREC_ACOND = 'SI'
  AND ALOJ.PISCINA = 'NO'
  AND ALOJ.SALON = 'SI'
  AND ALOJ.CANT_BANYOS > 1
  AND ALOJ.SUPERFICIE >= 80
  AND ALOJ.DIST_METRO >= 1
  AND ALOJ.DIST_CENTRO >= 2
  AND ALOJ.PRECIO_ALQUILER BETWEEN 1500 AND 2000
  AND ALOJ.PORCENTAJE_RESERVA = 25
  AND ALOJ.PUNTOS_AGENCIA > 4.5
  AND ALOJ.PUNTOS > 4) AS RES
  WHERE CANT_HAB >= 4
  ORDER BY PUNTOS DESC;
 """
pd.read_sql(text(sSQL), conn)
```

Le recomendaría el alojamiento ALOJ015 Dúplex Las Rocas porque, además de cumplir con todas las exigencias del cliente, es una de las pocas opciones que ofrece cuatro habitaciones, lo que garantiza una estancia más cómoda para un grupo de ocho personas, y entre las dos propiedades que cumplen todos los requisitos y cuentan con esa capacidad, resulta ser la alternativa más económica dentro del rango permitido y también la que presenta la mejor puntuación (No habiendo diferencia entre los puntos de confianza de la agencia, ambas con muy buen puntaje, teniendo en cuenta sus anteriores experiencias de intentos de estafas sufridos).

Responde las siguientes preguntas (con su respectiva query)

-¿Cuál es la suma total de metros cuadrados de los alojamientos que tienen piscina?

```python
sSQL = """
SELECT SUM(SUPERFICIE) AS SUMA_METROS_CUADRADOS
FROM alojamiento
WHERE piscina = "SI";
 """
pd.read_sql(text(sSQL), conn)
```

-¿Cuál es el alojamiento con mayor superficie?

-¿Cuál es el alojamiento con menor superficie?


```python
sSQL = """
SELECT CODIGO_ALOJAMIENTO, NOMBRE, SUPERFICIE
FROM alojamiento
ORDER BY SUPERFICIE ASC
LIMIT 1;
 """
pd.read_sql(text(sSQL), conn)
```

-¿Cuántos alojamientos tienen salón?


```python
sSQL = """
SELECT COUNT(SALON) AS CANTIDAD_DE_ALOJ_CON_SALON
FROM alojamiento
WHERE SALON = "SI";
 """
pd.read_sql(text(sSQL), conn)
```

-¿Cuántos alojamientos tienen terraza?

```python
sSQL = """
SELECT COUNT(TERRAZA) AS CANTIDAD_DE_ALOJ_CON_TERRAZA
FROM alojamiento
WHERE TERRAZA = "SI";
 """
pd.read_sql(text(sSQL), conn)
```

-¿Cuántos alojamientos hay por cada país según nuestra tabla de UBICACIÓN?

```python
sSQL = """
SELECT PAIS, COUNT(CODIGO_ALOJAMIENTO) AS CANTIDAD_DE_ALOJ_POR_PAIS
FROM UBICACION
GROUP BY PAIS;
 """
pd.read_sql(text(sSQL), conn)
```

-¿Cuál es el promedio de precios de alquiler de todos los alojamientos según la tabla PRECIO?


```python
sSQL = """
SELECT AVG(PRECIO_ALQUILER) AS PRECIO_PROMEDIO_POR_NOCHE
FROM PRECIO;
 """
pd.read_sql(text(sSQL), conn)
```

-¿Cuáles son las 3 agencias que tienen la mayor puntuación de agencia (campo a usar: PUNTOS_AGENCIA) según la tabla PUNTUACION?


```python
## EN ESTA QUERY SOLO TRAE TRES AGENCIAS PERO HAY 4 CON EL MISMO PUNTAJE
sSQL = """
SELECT AGENCIA, PUNTOS_AGENCIA
FROM PUNTUACION
ORDER BY PUNTOS_AGENCIA DESC
LIMIT 3
;
 """
pd.read_sql(text(sSQL), conn)
```

```python
## USANDO LA FUNCIÓN RANK()
sSQL = """
SELECT *
FROM (
    SELECT 
        AGENCIA,
        PUNTOS_AGENCIA,
        RANK() OVER (ORDER BY PUNTOS_AGENCIA DESC) AS rnk
    FROM puntuacion
) AS t
WHERE rnk <= 3
ORDER BY PUNTOS_AGENCIA DESC;

 """
pd.read_sql(text(sSQL), conn)
```

-Muestra los alojamientos (código y nombre) que: Contengan la palabra ‘Piso’ en su nombre o que tengan piscina y aire acondicionado, pero no pasen los 150 metros de superficie o que tengan más de un baño


```python
sSQL = """
SELECT CODIGO_ALOJAMIENTO, NOMBRE
FROM ALOJAMIENTO
WHERE (NOMBRE LIKE "%Piso%" OR (PISCINA = 'SI' AND AIREC_ACOND = 'SI'))
AND (SUPERFICIE <= 150 OR CANT_BANYOS > 1);
 """
pd.read_sql(text(sSQL), conn)
```

-Muestra una consulta que muestre el código de alojamiento, el nombre, el precio y una columna donde veamos el precio del alquiler con el símbolo del euro al final de cada importe, esta nueva columna se debe llama PRECIO_MODIF

```python
sSQL = """
SELECT 
    A.CODIGO_ALOJAMIENTO,
    A.NOMBRE,
    P.PRECIO_ALQUILER AS PRECIO,
    CONCAT(P.PRECIO_ALQUILER, ' €') AS PRECIO_MODIF
FROM ALOJAMIENTO AS A
INNER JOIN PRECIO AS P ON A.CODIGO_ALOJAMIENTO = P.CODIGO_ALOJAMIENTO;
 """
pd.read_sql(text(sSQL), conn)
```

-Muestra una columna sobre la tabla PRECIO que se llame CATEGORIA_PRECIO y que indique la palabra ‘Bajo’ si el precio está entre 780 y 1000, ‘Medio’ si el precio está entre 1000 y 1700 y ‘Alto’ si es mayor a 1700

```python
sSQL = """
SELECT 
    *,
    CASE WHEN PRECIO_ALQUILER BETWEEN 780 AND 1000 THEN "BAJO"
         WHEN PRECIO_ALQUILER BETWEEN 1000 AND 1700 THEN "MEDIO"
         WHEN PRECIO_ALQUILER > 1700 THEN "ALTO"
         END AS CATEGORIA_PRECIO   
FROM PRECIO;
 """
pd.read_sql(text(sSQL), conn)
```
